```python
metrics = [
    "cosine-Precision@3",
    "cosine-Recall@10",
    "cosine-MRR@10",
    "cosine-Accuracy@10",
]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()  # chuyển 2D → 1D cho dễ loop

for ax, metric in zip(axes, metrics):
    ax.plot(df["steps"], df[metric])
    ax.set_title(metric)
    ax.set_xlabel("Steps")
    ax.set_ylabel("Score")
    ax.grid(True)

fig.suptitle("Evaluation Metrics over Steps", fontsize=14)
plt.tight_layout(rect=[0, 0, 1, 0.96])  # chừa chỗ cho suptitle
plt.show()
```
